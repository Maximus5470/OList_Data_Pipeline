```python
%sql
create schema gold;
```

```python
%sql
drop view if exists gold.dim_products;

create view gold.dim_products as
select
    product_key as product_id,
    product_category_name as category,
    product_price as price,
    product_name_length as name_length,
    product_description_length as description_length,
    product_photos_qty as photos_qty,
    product_weight_g as weight_g,
    product_length_cm as length_cm,
    product_height_cm as height_cm,
    product_width_cm as width_cm
from silver.dim_products;
```

```python
%sql
drop view if exists gold.dim_reviews;

create view gold.dim_reviews as
select
    review_key as review_id,
    review_score as score,
    review_comment_title as review_title,
    review_comment_message as review_message,
    review_creation_date as creation_date,
    review_answer_timestamp as answer_date
from silver.dim_reviews
```

```python
%sql
drop view if exists gold.dim_customers;

create view gold.dim_customers as
select
    customer_key as customer_id,
    customer_unique_id as unique_id,
    customer_zip_code_prefix as zip_code_prefix,
    customer_city as city,
    customer_state as state
from silver.dim_customers
```

```python
%sql
drop view if exists gold.dim_sellers;

create view gold.dim_sellers as
select
    seller_key as seller_id,
    seller_zip_code_prefix as zip_code_prefix,
    seller_city as city,
    seller_state as state
from silver.dim_sellers
```

```python
%sql
drop view if exists gold.dim_payments;

create view gold.dim_payments as
select
    payment_key as payment_id,
    payment_sequential as sequential,
    payment_type,
    payment_installments as no_of_installments,
    payment_value
from silver.dim_payments
```

```python
%sql
drop view if exists gold.fact_orders;

create view gold.fact_orders as
select
    sales_key as sales_id,
    customer_key as customer_id,
    product_key as product_id,
    seller_key as seller_id,
    payment_key as payment_id,
    review_key as review_id,
    order_id,
    order_status,
    order_purchase_timestamp as purchase_timestamp,
    order_approved_at as approved_timestamp,
    order_delivered_carrier_date as carrier_date,
    order_delivered_customer_date as customer_date,
    order_estimated_delivery_date as estimated_delivery_date
from silver.fact_orders
```
