### Customer Dataset Table

```python
def read_table(path):
    df = spark.read.format('csv').options(header='true', inferSchema='true').load(path)
    df.limit(5).display()
    return df
```

```python
df = read_table("/Volumes/workspace/default/olist/olist_customers_dataset.csv")
```

```python
df.printSchema()
```

```python
df.createOrReplaceTempView("df")
```

customer_unique_id is not primary key

```python
%sql
select * 
from (select 
    customer_unique_id, 
    row_number() over(partition by customer_unique_id order by customer_unique_id) rn
from df)
where rn > 1
```

customer_id is the primary key

```python
%sql
select * 
from (select 
    customer_id, 
    row_number() over(partition by customer_id order by customer_id) rn
from df)
where rn > 1
```
