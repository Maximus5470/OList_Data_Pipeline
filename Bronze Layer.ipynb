```python
customers = spark.read.format("csv").options(header='true', inferSchema='true').load("/Volumes/workspace/default/olist/olist_customers_dataset.csv")
customers.write.format("delta").mode("overwrite").saveAsTable("bronze.customers")
```

```python
%sql
select * from bronze.customers
```

```python
%sql
select 
    product_id, 
    freight_value,
    row_number() over(partition by product_id order by price asc) as price_rank
from bronze.order_items
```
